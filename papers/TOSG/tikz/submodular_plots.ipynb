```python
def list_to_tikz_string(A):
    return '{' + str(A)[1:-1]+'}'

def line_to_tikz_string(Y):
    if len(Y) ==0:
        return ''
    result = ''
    y_last = 0
    for x,y in enumerate(Y):
        result += f'\\draw[-, blue] ({x}, {y_last}) -- ({x+1}, {y});'
        y_last = y
    result +=  f'\\node at({x+1}, {y}) [below right, blue, font=\\tiny] {{$x(t)$}};'
    return result

def modular_to_line(X, A, color):
    if len(X) == 0:
        return ''
    result = ''
    b_last = 0
    x_last = 0
    for t in range(4):
        if t+1 in A:
            b = b_last + X[t] - x_last
            result += f'\\draw[{color}, -] ({t}, {b_last}) -- ({t+1}, {b});'
            b_last = b
        else:
            result += f'\\draw[{color}, dashed] ({t}, {b_last}) -- ({t+1}, {b_last});'
        x_last = X[t]
    if color == 'green':
        result += f'\\node at ({t+1}, {b_last}) [above right, darkgreen, font=\\tiny] {{$b_T(A)$}};'
    else:
        result += f'\\node at ({t+1}, {b_last}) [above right, red, font=\\tiny] {{$p_T(A)$}};'

    return result



def modular_plot(X, A, x_lim, filename, show_x_tikz=True, supermodular=False):
    x_tikz = '{ }'
    color = 'green'
    if show_x_tikz: x_tikz = '{1,2,3,4}'
    if supermodular: color = 'red'
    # Define the TikZ LaTeX code
    tikz_code = f"""\
    \\begin{{tikzpicture}}
        % Draw x-axis

        \\draw[->] (-.5,0) -- (4.5,0) node[right, font=\\small] {{$t$}};
        \\draw[dashed, opacity=0.4] (0,{x_lim}) -- (4.5,{x_lim});
        \\node at (0, {x_lim}) [left, font=\\tiny] {{$\overline{{x}}$}};
        % Draw y-axis
        \\draw[->] (0,-2) -- (0,2.2);

        \\node at (0, 2) [left, font=\\tiny] {{$x(t)$}};

        % Add ticks on x-axis
        \\foreach \\x in {{1,2,3,4}}
            \\draw (\\x,0.1) -- (\\x,-0.1);
        
        \\foreach \\x in {x_tikz}
            \\draw (\\x - 0.5 ,0) -- (\\x-0.5,-0) node[below, font=\\tiny] {{\\x}};

        % Shade a vertical region
        \\foreach \\x in {list_to_tikz_string(A)}
            \\fill[{color}, opacity=0.2] (\\x-1,-2) rectangle (\\x,2);
        {line_to_tikz_string(X)}
        {modular_to_line(X, A, color)}
            
    \\end{{tikzpicture}}

    """

    # Write the TikZ code to a .tex file
    output_file = f"./figures/{filename}.tex"

    with open(output_file, "w") as file:
        file.write(tikz_code)

    print(f"TikZ code has been written to {output_file}")
```
